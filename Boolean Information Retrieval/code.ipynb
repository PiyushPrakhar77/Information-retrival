```python
import re
```

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)  # remove special characters
    words = text.split()
    return words
```

```python
def build_inverted_index(documents):
    inverted_index = {}

    for doc_id, content in documents.items():
        words = preprocess_text(content)

        for word in words:
            if word not in inverted_index:
                inverted_index[word] = set()
            inverted_index[word].add(doc_id)

    return inverted_index
```

```python
def boolean_query(query, inverted_index, documents):
    words = query.lower().split()
    result = set(documents.keys())

    i = 0
    while i < len(words):

        if words[i] == "and":
            term = words[i + 1]
            result = result & inverted_index.get(term, set())
            i += 2

        elif words[i] == "or":
            term = words[i + 1]
            result = result | inverted_index.get(term, set())
            i += 2

        elif words[i] == "not":
            term = words[i + 1]
            result = result - inverted_index.get(term, set())
            i += 2

        else:
            result = inverted_index.get(words[i], set())
            i += 1

    return result
```

```python
documents = {
    1: "The quick brown fox jumps over the lazy dog.",
    2: "The lazy dog lies in the sun.",
    3: "A fox is quick and clever.",
    4: "Sunshine makes the fox happy."
}
```

```python
inverted_index = build_inverted_index(documents)

print("Inverted Index:\n")
for term, doc_ids in inverted_index.items():
    print(term, ":", doc_ids)
```

```python
print("\nBoolean Query Results:\n")

query1 = "fox AND quick"
print(query1, "=>", boolean_query(query1, inverted_index, documents))

query2 = "fox AND NOT lazy"
print(query2, "=>", boolean_query(query2, inverted_index, documents))

query3 = "dog OR  sunshine"
print(query3, "=>", boolean_query(query3, inverted_index, documents))
```
