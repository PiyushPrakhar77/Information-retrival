```python
import re
from collections import defaultdict

def build_comprehensive_positional_index(documents):
    positional_index = defaultdict(lambda: defaultdict(list))

    for doc_id, doc in enumerate(documents)
        words = re.findall(r'\w+', doc.lower())

        for position, word in enumerate(words):
            positional_index[word][doc_id].append(position)

    return dict(positional_index)
```

```python
Documents = [
    "The quick brown fox jumps over the lazy dog.",
    "The lazy dog sleeps.",
    "A quick brown dog outpaces a quick fox"
]

positional_index = build_comprehensive_positional_index(Documents)

for term, posting in sorted(positional_index.items()):
    total_count = sum(len(positions) for positions in posting.values())

    formatted_posting = {
        doc_id + 1: [pos + 1 for pos in positions]
        for doc_id, positions in posting.items()
    }

    print(f"{term} : count={total_count}, docs={formatted_posting}")
```

```python
def phrase_query_with_positions(positional_index, phrase):
    terms = phrase.lower().split()

    if any(term not in positional_index for term in terms):
        return {}

    # candidate documents
    candidate_docs = set(positional_index[terms[0]].keys())
    for term in terms[1:]:
        candidate_docs &= set(positional_index[term].keys())

    result = {}

    for doc_id in candidate_docs:
        positions = positional_index[terms[0]][doc_id]

        for i in range(1, len(terms)):
            next_positions = positional_index[terms[i]][doc_id]
            positions = [p + 1 for p in positions if p + 1 in next_positions]

        if positions:
            # convert to 1-based positions
            result[doc_id + 1] = [p + 1 for p in positions]

    return result
```

```python
phrase = input("Enter phrase query: ")

matches = phrase_query_with_positions(positional_index, phrase)

if matches:
    print("Phrase found at:")
    for doc_id, pos in matches.items():
        print(f"Document {doc_id} → position(s): {pos}")
else:
    print("Phrase not found in any document.")
```
